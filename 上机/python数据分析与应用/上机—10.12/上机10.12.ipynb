```python
from idlelib.iomenu import encoding

import pandas as pd
from holoviews.operation import transform
from pandas.conftest import unit

df = pd.read_csv("某地区房屋销售数据.csv", skiprows=1, names=["房屋售出时间","地区邮编","房屋价格（元）","房屋类型","配套房间数（间）"])
print(df.head())
print(df.columns)
```

```python
print("数据维度ndim：", df.ndim)
print("数据形状shape(行数,列数)：", df.shape)
print("所有特征名称columns：")
print(df.columns)
```

```python
df_loc = df.loc[df["房屋类型"] == "unit"]
df_loc.head()
```

```python
# 布尔条件：标记哪些行是unit
unit_room=df["房屋类型"]=="unit"
# 取出条件为True（房屋类型=unit）的行索引
df_p=unit_room[unit_room].index
# iloc根据筛选后的行索引提取数据
df_iloc=df.iloc[df_p,:]
print(df_iloc.head())
```

```python
df["房屋售出时间"]=pd.to_datetime(df["房屋售出时间"])
print(df["房屋售出时间"].head())
```

```python
mean_price = df["房屋价格（元）"].mean()
max_price = df["房屋价格（元）"].max()
min_price = df["房屋价格（元）"].min()
print("房屋价格均值：",mean_price)
print("房屋价格最大值：",max_price)
print("房屋价格最小值：",min_price)
```

```python
p_dec=df["房屋价格（元）"].describe()
print(p_dec)
```

```python
df["new_postcode"]=df["地区邮编"].apply(lambda x:str(x)[0:2])
print(df[["new_postcode"]].head())
print(df[["地区邮编","new_postcode"]].head())
```

```python
group_count = df.groupby("new_postcode").agg(房屋售出总数=("地区邮编","count"))
print("各地区房屋售出总数：")
print(group_count)
```

```python
housesale1 = df.groupby(["房屋类型","new_postcode"]).count()
print("housesale1分组结果：")
print(housesale1.head())
```

```python
housesale1 = df.groupby(["房屋类型","new_postcode"])
df["房屋价格分组均值"] = housesale1["房屋价格（元）"].transform("mean")
print(df[["房屋类型","new_postcode","房屋价格（元）","房屋价格分组均值"]].head())
```

```python
import pandas as pd
df_stroke = pd.read_excel(r"C:\Users\ASUS\Downloads\healthcare-dataset-stroke.xlsx")
print("stroke数据集前5行：")
print(df_stroke.head())
```

```python
df_age = pd.read_excel(r"C:\Users\ASUS\Downloads\healthcare-dataset-age_abs.xlsx")
print("age_abs数据集前5行：")
print(df_age.head())
```

```python
print("age量表形状(行数,列数)：", df_age.shape)
print(df_age.info())
```

```python
df_merge = pd.merge(df_stroke, df_age, on="编号", how="outer")
print("合并后数据预览：")
print(df_merge.head())
```

```python
print("合并后：",df_merge.info())
print(df_merge.head())
print(df_merge.columns)
print(df_merge.index)
```

```python
age_series = df_merge["年龄/岁"]
print("年龄字段前10个值：")
print(age_series.head(10))
```

```python
is_abnormal = []
for val in df_merge["年龄/岁"]:
    if pd.notna(val):
        if isinstance(val, float) and not val.is_integer():
            is_abnormal.append(True)
        else:
            is_abnormal.append(False)
    else:
        is_abnormal.append(False)
df_merge["异常年龄"] = is_abnormal
print("标记后的异常判断结果：")
print(df_merge[["年龄/岁","异常年龄"]].head(10))
```

```python

df_clean = df_merge[df_merge["异常年龄"] == False]
print("删除异常值之后数据shape：",df_clean.shape)
print(df_clean[["年龄/岁"]].head())
```

```python
age_col = df_clean["年龄/岁"]
print("清洗后的年龄数据：")
print(age_col.head())
```

```python
df_clean["wid_age"]=pd.cut(age_col,bins=4)
print("等宽处理后:")
print(df_clean[["年龄/岁","wid_age"]].head(10))
```
