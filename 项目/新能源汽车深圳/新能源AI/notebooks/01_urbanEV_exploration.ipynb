```python
import pandas as pd
df = pd.read_csv("../dataset/raw/occupancy.csv")
df
```

```python
df.shape
```

```python
df.info()
```

```python
occupancy_long = df.melt(
    id_vars=["time"],
    var_name="zone",
    value_name="occupancy"
)

occupancy_long.head()
```

```python
occupancy_long.shape
```

```python
occupancy_long.isnull().sum()
```

```python
zone_avg = (
    occupancy_long
    .groupby("zone")["occupancy"]
    .mean()
    .sort_values(ascending=False)
)

zone_avg.head(10)
```

```python
occupancy_long["occupancy"].describe()
```

```python
occupancy_long[occupancy_long["occupancy"]>100].head()
```

```python
zone_overload = (
    occupancy_long
    .assign(overload=occupancy_long["occupancy"] > 100)
    .groupby("zone")["overload"]
    .mean()
    .sort_values(ascending=False)
)

zone_overload.head(10)
```
